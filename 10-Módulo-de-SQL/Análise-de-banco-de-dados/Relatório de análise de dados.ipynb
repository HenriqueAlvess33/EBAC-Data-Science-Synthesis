## Existem valores discrepantes (outliers) nos dados?

### **Método MAD (Median Absolute Deviation)**

#### idade

> ```sql
> WITH 
> -- Calcular mediana
> ordered AS (
>     SELECT idade,
>            ROW_NUMBER() OVER (ORDER BY idade) as row_num,
>            COUNT(*) OVER () as total_count
>     FROM credito
> ),
> mediana AS (
>     SELECT AVG(idade) as mediana
>     FROM ordered
>     WHERE row_num IN ((total_count + 1) / 2, (total_count + 2) / 2)
> ),
> -- Calcular MAD
> desvios AS (
>     SELECT ABS(c.idade - m.mediana) as desvio_absoluto
>     FROM credito c, mediana m
> ),
> ordered_desvios AS (
>     SELECT desvio_absoluto,
>            ROW_NUMBER() OVER (ORDER BY desvio_absoluto) as row_num,
>            COUNT(*) OVER () as total_count
>     FROM desvios
> ),
> mad_calc AS (
>     SELECT AVG(desvio_absoluto) as mad
>     FROM ordered_desvios
>     WHERE row_num IN ((total_count + 1) / 2, (total_count + 2) / 2)
> )
> -- Identificar outliers
> SELECT 
>     c.*,
>     ROUND(ABS(c.idade - m.mediana) / mad.mad, 2) as mad_score,
>     CASE 
>         WHEN ABS(c.idade - m.mediana) / mad.mad > 3.5 THEN 'OUTLIER_SEVERO'
>         WHEN ABS(c.idade - m.mediana) / mad.mad > 2.5 THEN 'OUTLIER_MODERADO'
>         ELSE 'NORMAL'
>     END as classificacao
> FROM credito c, mediana m, mad_calc mad
> ORDER BY mad_score DESC;
> ```
>

### **Qual o nível de correlação entre os dados ?**

````python
# Importação das bibliotecas necessárias
import pandas as pd  # Para manipulação de dados
import matplotlib.pyplot as plt  # Para criação de gráficos
import seaborn as sns  # Para visualizações estatísticas mais avançadas

# Carregamento do dataset a partir de um arquivo CSV
df = pd.read_csv("credit_database.csv")

# Renomeação das colunas para português para melhor entendimento
df.columns = [
    'idade',                # Idade do cliente (em anos)
    'sexo',                 # Gênero do cliente
    'dependentes',          # Número de dependentes do cliente
    'escolaridade',         # Nível de escolaridade
    'estado_civil',         # Estado civil do cliente
    'salario_anual',        # Renda anual do cliente
    'tipo_cartao',          # Categoria do cartão de crédito
    'qtd_produtos',         # Quantidade de produtos contratados
    'iteracoes_12m',        # Número de interações nos últimos 12 meses
    'meses_inativo_12m',    # Meses inativo nos últimos 12 meses
    'limite_credito',       # Limite de crédito disponível
    'valor_transacoes_12m', # Valor total das transações nos últimos 12 meses
    'qtd_transacoes_12m'    # Quantidade de transações nos últimos 12 meses
]

# Criação da matriz de correlação apenas com variáveis numéricas
# select_dtypes seleciona apenas colunas do tipo float64 e int64 (numéricas)
# corr() calcula a correlação de Pearson entre todas as variáveis numéricas
corr_matrix = df.select_dtypes(include=['float64', 'int64']).corr()

# Configuração do gráfico de heatmap (mapa de calor)
plt.figure(figsize=(10, 8))  # Define o tamanho da figura como 10x8 polegadas

# Cria o heatmap da matriz de correlação com as seguintes configurações:
# - annot=True: Mostra os valores de correlação dentro de cada célula
# - cmap='coolwarm': Usa escala de cores do azul (negativo) ao vermelho (positivo)
# - center=0: Centraliza a escala de cores no zero (cor neutra)
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0)

# Adiciona título ao gráfico
plt.title('Matriz de Correlação - Variáveis do Cartão de Crédito')

# Exibe o gráfico
plt.show()

# ANÁLISE SUGERIDA:
# - Valores próximos de +1: correlação positiva forte (quando uma variável aumenta, a outra também aumenta)
# - Valores próximos de -1: correlação negativa forte (quando uma variável aumenta, a outra diminui)
# - Valores próximos de 0: pouca ou nenhuma correlação linear
# - Procure por correlações fortes que possam indicar relações importantes no negócio
````

## Explanação do dataset

* Levando em consideração que a correlação entre idade e escolaridade indicada no diagrama é 0. Qual a explicação para esta falta de encaixe entre as variáveis ?

````python
# Carregando o dataset
url = 'https://raw.githubusercontent.com/HenriqueAlvess33/EBAC-Data-Science-Synthesis/refs/heads/main/Data/Arquivos%20CSV/credit_database.csv'
df = pd.read_csv(url)
df.columns = ['idade', 'sexo', 'dependentes', 'escolaridade', 'estado_civil', 'salario_anual', 'tipo_cartao', 'qtd_produtos', 'iteracoes_12m', 'meses_inativo', 'limite_credito', 'valor_transacoes_12m', 'qtd_transacoes_12m']

# Cria uma cópia do DataFrame original para evitar modificar os dados de origem
df_visualizacao_idade = df.copy()

# Cria faixas de idade divididas em 4 quartis usando qcut
df_visualizacao_idade['faixas_idade'] = pd.qcut(df_visualizacao_idade['idade'], 4)

# Ordena as faixas de idade pelo limite inferior de cada intervalo
faixas_idade = sorted(
    df_visualizacao_idade.faixas_idade.unique().tolist(),
    key=lambda x: x.left
)

# Cria subconjuntos do DataFrame, um para cada quartil de idade
Q1_idade = df_visualizacao_idade[df_visualizacao_idade['faixas_idade'] == faixas_idade[0]]
Q2_idade = df_visualizacao_idade[df_visualizacao_idade['faixas_idade'] == faixas_idade[1]]
Q3_idade = df_visualizacao_idade[df_visualizacao_idade['faixas_idade'] == faixas_idade[2]]
Q4_idade = df_visualizacao_idade[df_visualizacao_idade['faixas_idade'] == faixas_idade[3]]

# Lista com os DataFrames dos quartis para facilitar iteração
quarters = [Q1_idade, Q2_idade, Q3_idade, Q4_idade]

# Cria rótulos descritivos para cada quartil, incluindo o intervalo da faixa
quarter_labels = [f"Q{i+1}: {faixa}" for i, faixa in enumerate(faixas_idade)]

# Exibe estatísticas básicas dos quartis
print("Estatísticas por quartil de idade:")

for i, (quarter, label) in enumerate(zip(quarters, quarter_labels)):
    total = len(quarter)
    print(f"\n{label}:")
    print(f"  Total de pessoas: {total}")
    print(f"  Distribuição de escolaridade:")

    # Calcula frequência de cada nível de escolaridade
    educ_counts = quarter['escolaridade'].value_counts()

    # Exibe contagem e porcentagem de cada categoria
    for educ, count in educ_counts.items():
        percentage = (count / total) * 100
        print(f"    {educ}: {count} ({percentage:.1f}%)")

# Cria figura com 4 gráficos (2 linhas x 2 colunas)
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.ravel()  # Facilita a indexação linear dos eixos

# Gera um gráfico de contagem para cada quartil
for i, (quarter, label) in enumerate(zip(quarters, quarter_labels)):

    # Gráfico de barras mostrando distribuição da escolaridade no quartil
    sns.countplot(
        x='escolaridade',
        data=quarter,
        order=quarter['escolaridade'].value_counts().index,  # Ordena pelas categorias mais frequentes
        ax=axes[i]
    )

    # Título com nome do quartil + tamanho da amostra
    axes[i].set_title(f'Q{i+1}: {label}\n(n={len(quarter)})')

    # Rotaciona rótulos do eixo X para melhor leitura
    axes[i].tick_params(axis='x', rotation=45)

    # Adiciona rótulos de valor acima das barras
    for p in axes[i].patches:
        axes[i].annotate(
            f'{p.get_height():.0f}',
            (p.get_x() + p.get_width() / 2., p.get_height()),
            ha='center', va='bottom'
        )

# Ajusta espaçamento da figura para evitar sobreposição
plt.tight_layout()

# Exibe todos os gráficos
plt.show()
````

- **O "Piso" do VIP sobe:** Para Gold e Silver, o limite mínimo acompanha o salário. Um cliente Gold que ganha 80k já entra no banco com um limite garantido muito maior (aprox. 34k no gráfico).

## **Relação Quantidade de Transação x Tipo Cartão**

### Mínima

- **Hierarquia Definida:** Existe uma escada clara no *piso* de uso. Quem tem cartões "superiores" (Gold e Silver) tende a fazer um número mínimo de transações maior do que quem tem o cartão de entrada (Blue).
- **Gold (Laranja):** É o cartão com o maior "engajamento mínimo". Mesmo quem usa pouco o Gold, usa mais que o Silver ou Blue.
- **Blue (Azul):** O mínimo é baixo e constante (~10 transações) em todas as faixas de renda. Isso condiz com um cartão de entrada/massificado.

### Média

- **Anomalia no Gold (60K - 80K):** Há um pico gigantesco na média de transações do cartão Gold na faixa salarial de 60K-80K (chegando a quase 60 transações).
    - *Hipótese:* Pode haver um grupo específico de clientes "heavy users" nesta faixa que puxa a média para cima, ou uma campanha de incentivo específica para este perfil.
- **Estabilidade do Blue:** A média do cartão Blue é uma linha quase reta (~40 transações). Como a base de clientes é enorme, a média é muito estável (lei dos grandes números). A renda do cliente não altera o quanto ele usa o cartão Blue em média.
- **Queda nos Ricos ($120K+):** Curiosamente, na faixa de renda mais alta, a média de transações cai para quase todos os cartões (exceto Silver, que se mantém).

### Máxima

- **Inversão de Papéis:** Ao contrário da média e da mínima, o cartão **Blue** possui os maiores picos de uso (máximas chegando a 100 transações).
    - *Interpretação:* Embora a *média* do Blue seja baixa, existem "super usuários" (outliers) dentro da categoria Blue que transacionam muito, superando até os usuários de Gold e Silver.
- **Instabilidade do Gold:** O cartão Gold mostra um comportamento errático. Cai bruscamente na faixa 40K-60K e explode na faixa 60K-80K (corroborando o gráfico de média).

- **O Comportamento do "Orange" na faixa 60k-80k:** Este é o ponto mais chamativo dos dados. Clientes com renda média alta (60-80k) que possuem o cartão **Orange**são os mais engajados da base (maior média e picos altos). Vale investigar o porquê: *Benefícios específicos? Perfil de gastos focado em acúmulo de pontos?*
- **O Cartão Blue é "Para Todos":** Ele tem a maior base, comportamento estável na média e abrange desde quem usa muito pouco até quem usa excessivamente (máximas altas). A renda do cliente não dita o quanto ele usa o cartão Blue.
- **Limite de Dados:** As conclusões sobre o cartão **Platinum** (verde) devem ser descartadas ou tratadas com muito cuidado, pois não há pontos de dados suficientes para traçar uma linha contínua.
- **Renda vs. Uso:** Não há uma correlação linear direta onde "quanto maior a renda, mais transações". Pelo contrário, a faixa de renda mais alta ($120K+) tende a ter um volume de transações menor ou igual às faixas médias.

## Meses inativos x Tipo Cartão

### 1. Visão Geral: Estabilidade e o "Teto" de 6 Meses

De modo geral, a média de inatividade gira em torno de **2 a 2,5 meses**. Isso indica que a base de clientes é razoavelmente ativa.

- **O Teto de 6 Meses (Gráfico Máxima):** Note que a linha azul (Blue) crava no valor **6** em todas as faixas salariais. Isso sugere que "6 meses" é provavelmene o ponto de corte da coleta de dados ou o gatilho para o banco considerar o cliente como *churn* (perdido). O cartão Blue é o único que atinge esse teto consistentemente.

### 2. Análise por Tipo de Cartão

### **Cartão Blue (A Massa Estável)**

- **Comportamento:** É extremamente previsível. A média de inatividade (linha azul no gráfico superior direito) é quase uma reta constante, variando muito pouco (entre 2.2 e 2.3 meses), independentemente se o cliente ganha menos de 40k ou mais de 120k.
- **Risco:** Como visto no gráfico de "Máxima", é no cartão Blue que estão concentrados os clientes que ficam o tempo máximo (6 meses) sem usar o cartão. **A renda não influencia a fidelidade/atividade do cliente Blue.**

### **Cartão Gold (O Comportamento Curioso nos Ricos)**

- **A Queda Brusca (Gráfico Média):** O dado mais interessante está no canto direito do gráfico de média. Para clientes com renda superior a **$120K+**, a inatividade do cartão Gold despenca para **1.0 mês**.
    - *Interpretação:* Clientes ricos com cartão Gold são os mais ativos de toda a base. Eles praticamente não deixam o cartão parado.
- **Alerta de Viés (Gráfico Contagem):** Olhe para o gráfico de contagem (inferior direito). A quantidade de clientes Gold na faixa de $120k+ é muito baixa (linha laranja próxima a zero). **Cuidado:** Essa média excelente de "1 mês inativo" pode ser resultado de uma amostra muito pequena (ex: 5 clientes que usam muito o cartão), não sendo estatisticamente representativa para uma regra geral.

### **Cartão Silver (Oscilação)**

- Apresenta uma leve piora na inatividade (sobe para quase 2.4 meses na média) nas faixas de renda média-alta (60k-120k), mas nada alarmante. Ele flutua mais que o Blue, mas não tem os extremos do Gold.

### 3. Síntese dos Insights

1. **O "Peso Morto" está no Blue:** Os clientes que abandonam o cartão (ficam 6 meses inativos - gráfico Máxima) são predominantemente usuários do cartão Blue. Isso reforça a necessidade de campanhas de reativação focadas especificamente nesse segmento, já que é o maior volume do banco.
2. **Gold em $120k+ é um Nicho de Ouro (com ressalvas):** Aparentemente, quem ganha muito bem e opta pelo Gold é extremamente fiel/ativo. Vale investigar por que esses clientes de alta renda não estão no Platinum e se esse comportamento se sustenta se a base de usuários aumentar.
3. **Independência da Renda:** Para a grande maioria (Cartão Blue), ganhar mais ou menos não afeta a frequência de uso. A inatividade é uma questão de comportamento/produto, não de capacidade financeira.

# Conclusão Geral e Insights Estratégicos

Após a análise exploratória dos dados, consolidamos os principais achados em quatro pilares fundamentais, seguidos de um plano de ação recomendado.

---

### 1. Perfil Demográfico e Estrutural

> Este dataset reflete um nicho específico, não a população geral. Há um viés de seleção por nível socioeconômico.
> 
- **Homogeneidade Educacional:** Predominância de educação superior (mestrado), independente da idade.
- **Ciclo de Vida:** Padrão clássico com pico de dependentes na meia-idade (45-51 anos) e redução após os 50.
- **Gênero:** Base majoritariamente masculina (61%), mas com comportamentos distintos entre os grupos.

### 2. Disparidade de Gênero (Critical Finding)

Identificamos uma assimetria sistêmica alarmante:

- **Teto de Crédito:** Homens dominam as faixas superiores (88% dos clientes acima de R$ 16k são homens).
- **Piso Salarial:** Mulheres com mestrado concentram-se na faixa salarial mais baixa (< $40K).
- **Impacto Social:** Mulheres divorciadas possuem mais dependentes (mediana: 3) que homens na mesma situação (mediana: 2).
- **Eficiência:** Mulheres transacionam mais vezes, mesmo com limites menores.

### 3. Hierarquia de Produtos

Os cartões apresentam dinâmicas de engajamento distintas:

| Produto | Perfil e Comportamento |
| --- | --- |
| **Blue** | *Massa.* Maior base, abrange de casuais a "super-users". Alto risco de inatividade (6 meses). |
| **Gold** | *Estratégico.* Comportamento errático. Pico de engajamento na faixa salarial 60k-80k. |
| **Silver** | *Intermediário.* Comportamento oscilante sem padrão claro. |

**Nota:** Observou-se um "Teto Universal" (~34.5k) que iguala todos os cartões para clientes de alta renda.

### 4. Economia do Cliente: Renda vs. Crédito

*"Renda determina potencial, mas não realizações."*

- **Correlação Incompleta:** Salário maior tende a limite médio maior, mas o piso mínimo permanece constante.
- **Segmentação:** O tipo de cartão define "limites base" que independem da renda.
- **Volume:** Alto volume transacional não garante limite alto.
