```python
import pandas as pd  # Importa a biblioteca pandas para manipulação de dados
import os  # Importa a biblioteca os para operações com o sistema de arquivos

# Lê o arquivo CSV e armazena o conteúdo em um DataFrame
df = pd.read_csv('./Base de dados/online_shoppers_intention.csv')

# Realiza uma amostragem aleatória de 1000 linhas do DataFrame
df_amostragem = df.sample(n=1000)

# Cria o diretório 'output' caso ele não exista
os.makedirs("./output", exist_ok=True)

# Salva o DataFrame amostrado em um novo arquivo CSV dentro do diretório 'output'
df_amostragem.to_csv('./output/df_amostragem.csv')
```
