```python
# Importa o módulo NumPy para operações matemáticas e manipulação de arrays
import numpy as np

# Importa o Matplotlib para criação de gráficos
import matplotlib.pyplot as plt

# Importa o módulo warnings para gerenciar mensagens de aviso
import warnings
```

```python
# Ignora os avisos para evitar mensagens desnecessárias
warnings.filterwarnings("ignore")

# Define a função f(x) como x^3
f = lambda x: np.power(x, 3)

# Função para calcular a área aproximada usando a soma de Riemann
def calcula_area(a, b, func, frame_num):
    # Define o número de divisões (retângulos)
    bins = frame_num + 1
    # Calcula o tamanho de cada divisão (delta)
    delta = (b - a) / bins
    # Gera os pontos médios das colunas (retângulos)
    # x_cols = np.linspace(a, b, bins) + delta / 2
    x_cols = np.linspace(a + delta/2, b - delta/2, bins)

    # Calcula os valores de f(x) nos pontos médios
    ycols = f(x_cols)
    
    # Retorna a soma das áreas dos retângulos
    return sum(delta * ycols)

# Função para animar a aproximação de Riemann
def animate(n, a=1, b=2, func=f):
    # Limpa o gráfico atual
    plt.cla()
    plt.close()
    # Cria uma nova figura e eixos para o gráfico
    fig, ax = plt.subplots(figsize=(10, 5))
    # Define os limites do gráfico no eixo x e y
    ax.set_xlim(a, b)
    ax.set_ylim(0, 20)
    # Define o título do gráfico
    plt.title("Aproximação de Riemann")
    
    # Gera os pontos para desenhar a curva da função
    x = np.linspace(a, b, 1000)
    y = f(x)
    # Plota a curva da função
    ax.plot(x, y, "-", color='blue', label='f(x)')
    
    # Configura os rótulos do eixo x
    ax.tick_params(axis='x', labelsize=10, rotation=45)
    # Adiciona uma grade ao gráfico
    ax.grid()
    
    # Define o número de divisões (retângulos)
    bins = n + 1
    # Calcula o tamanho de cada divisão (delta)
    delta = (b - a) / bins
    # Gera os pontos médios das colunas (retângulos)
    # x_cols = np.linspace(a, b, bins) + delta / 2
    x_cols = np.linspace(a + delta/2, b - delta/2, bins)
    # Calcula os valores de f(x) nos pontos médios
    ycols = f(x_cols)
    
    # Desenha os retângulos da soma de Riemann
    bar = ax.bar(x_cols, ycols, width=delta, align='center', alpha=0.5, color='red', label='Aproximação de Riemann')
    # Calcula a área aproximada
    area = calcula_area(a, b, func, n)
    
    # Atualiza o título do gráfico com a área calculada
    ax.set_title(f"Aproximação de Riemann\nÁrea: {area:.4f} ")
    
    # Retorna os retângulos desenhados
    return bar

# Chama a função animate para gerar a aproximação com 35 divisões
animate(36)
```

```python
# Itera sobre uma lista de valores de n, variando de 0 a 35 com incrementos de 5
for n in range(0, 106, 5):
    # Calcula a área aproximada da integral usando a soma de Riemann
    # com o número de divisões especificado por n
    area = calcula_area(1, 2, f, n)
    # Exibe o valor da área calculada para cada valor de n
    print(f"Para n = {n}, a área aproximada é: {area:.4f}")
```
