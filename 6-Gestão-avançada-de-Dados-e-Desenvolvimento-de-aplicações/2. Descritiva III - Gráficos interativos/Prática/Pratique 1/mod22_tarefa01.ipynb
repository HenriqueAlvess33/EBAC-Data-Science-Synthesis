```python
# Importando bibliotecas para manipulação de dados
import pandas as pd

# Importando biblioteca para download de dados financeiros
import yfinance as yf

# Importando bibliotecas para visualização de dados
import plotly.express as px
from plotly import graph_objects as go

# Importando biblioteca para manipulação de arquivos e diretórios
import os
```

#### 1. Baixando dados de ativos
    1. Baixe os dados de três ativos da sua escolha do Yahoo Finance, correspondente a um período de pelo menos 2 anos até a data mais recente.
    2. Salve esses dados para que você possa utilizá-los na próxima tarefa.

```python
# Definindo a lista de ativos que serão analisados
ativos = ['INTC', 'AMD', 'IBM']

# Definindo a data inicial para o período de análise
data_inicial = '2023-04-15'

# Baixando os dados históricos dos ativos selecionados a partir da data inicial usando a biblioteca yfinance
df_ativos = yf.download(ativos, start=data_inicial)

# Removendo os nomes dos níveis de coluna do DataFrame para simplificar a manipulação dos dados
df_ativos.columns.names = [None, None]

# Exibindo o DataFrame com os dados baixados
df_ativos
```

#### 2. Observando a volatilidade dentro do dia
1. Calcule a amplitude de variação do valor do ativo dentro do dia. Essa amplitude você pode calcular com um comando simples de adicionar uma coluna no Pandas: (valor máximo - valor mínimo).
2. Monte um gráfico de linha mostrando a variação dessa amplitude ao longo do período coletado, por ativo.
3. Veja se consegue identificar que ativo apresentou maior volatilidade.

```python
# Selecionando as colunas 'High' e 'Low' do DataFrame df_ativos
# Empilhando os dados para transformar o DataFrame em um formato longo (long format)
# Resetando o índice para que as colunas do índice original se tornem colunas regulares
temp_df = df_ativos[['High', 'Low']].stack().reset_index()
```

```python
# Calculando a volatilidade diária para cada ativo
temp_df['Volatilidade'] = temp_df['High'] - temp_df['Low']  # Subtraindo o valor mínimo (Low) do valor máximo (High) para obter a amplitude diária
temp_df['Volatilidade'] = temp_df['Volatilidade'].abs()  # Garantindo que a volatilidade seja sempre positiva
temp_df['Volatilidade'] = temp_df['Volatilidade'].round(2)  # Arredondando os valores da volatilidade para duas casas decimais

# Renomeando a coluna 'level_1' para 'Ticker' para melhor identificação dos ativos
temp_df.rename(columns={'level_1': 'Ticker'}, inplace=True)

# Exibindo o DataFrame atualizado
temp_df
```

```python
# Extraindo apenas os valores High e Low
# Seleciona as colunas 'High' e 'Low' do DataFrame 'df_ativos' para os cálculos de volatilidade
high = df_ativos['High']
low = df_ativos['Low']

# Calculando a volatilidade
# Subtrai os valores mínimos (Low) dos valores máximos (High) para calcular a amplitude diária
# Aplica o valor absoluto para garantir que a volatilidade seja positiva e arredonda para duas casas decimais
volatilidade = (high - low).abs()

# Remove o nome do nível de coluna, caso exista, para simplificar a estrutura do DataFrame
volatilidade.columns.name = None
```

```python
# Exibindo o DataFrame 'volatilidade'
# Este DataFrame contém a volatilidade diária calculada para cada ativo (AMD, IBM, INTC),
volatilidade
```

```python
# Recriando as colunas multi-index corretas para a volatilidade
# Aqui, as colunas do DataFrame 'volatilidade' são convertidas para um formato de MultiIndex,
# onde o nível superior é 'Volatilidade' e o nível inferior contém os nomes das colunas originais (AMD, IBM, INTC).
volatilidade.columns = pd.MultiIndex.from_product([['Volatilidade'], volatilidade.columns])

# Concatenando o DataFrame 'volatilidade' com o DataFrame 'df_ativos'
# Isso adiciona as colunas de volatilidade ao DataFrame principal 'df_ativos', alinhando os índices.
df_ativos = pd.concat([df_ativos, volatilidade], axis=1)

# Ordenando as colunas do DataFrame 'df_ativos', se necessário
# Isso organiza as colunas em ordem alfabética ou lógica, dependendo da estrutura do MultiIndex.
df_ativos = df_ativos.sort_index(axis=1)
```
