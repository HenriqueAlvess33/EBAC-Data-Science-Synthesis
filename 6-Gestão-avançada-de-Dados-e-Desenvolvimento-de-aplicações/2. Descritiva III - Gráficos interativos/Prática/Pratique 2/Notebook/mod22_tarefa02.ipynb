#### 2. Média móvel

1. Faça uma média móvel  em 90 dias dos valores de fechamento ajustados dos ativos que você escolheu
2. Utilizando a fução ```rolling()```, calcule o desvio padrão desses ativos ao longo do tempo, utilizando a mesma janela móvel do item acima
2. Monte um gráfico de linhas para as médias móveis dos ativos da sua base
4. Monte um gráfico de linhas para os desvios padrão em janela móvel
5. Compare os dois gráficos que você fez com o da tarefa anterior. Veja se eles apontam conclusões semelhantes quanto aos períodos de maior volatilidade (variação) dos ativos.

````python
df
````

````python
df_mv_90 = df.Close.rolling(window=90).mean()
df_mv_90 = df_mv_90.stack().reset_index()
df_mv_90.columns = ["Data", 'Ativo',"Média Móvel 90 dias"]
df_mv_90
````

````python
fig = px.line(df_mv_90, x='Data', y='Média Móvel 90 dias', color='Ativo', 
			  title='Média Móvel de 90 dias',
              labels={'Data': 'Data', 'Média Móvel 90 dias': 'Média Móvel 90 dias'}, 
			  markers=True, 
			  template='plotly_white')
fig.update_traces(mode='lines+markers')
fig.show()
````

````python
df_std_90 = df.Close.rolling(window=90).std()
df_std_90 = df_std_90.stack().reset_index()
df_std_90.columns = ["Data", 'Ativo',"Desvio Padrão 90 dias"]
df_std_90
````

````python
fig = px.line(df_std_90, x='Data', y='Desvio Padrão 90 dias', color='Ativo', 
			  title='Desvio Padrão 90 dias',
              labels={'Data': 'Data', 'Desvio Padrão 90 dias': 'Desvio Padrão 90 dias'}, 
			  markers=True, 
			  template='plotly_white')
fig.update_traces(mode='lines+markers')
fig.show()
````

Resposta: O investimento mais consistente e de menor risco, neste caso, seria a Intel (INTC), que apresenta baixa volatilidade e comportamento estável ao longo do tempo. A AMD, por outro lado, demonstra alta volatilidade e uma tendência de queda nos últimos meses, o que indica maior risco. Já a IBM exibe uma trajetória de crescimento gradual e consistente, com níveis de volatilidade relativamente baixos na maior parte do período, mas com picos recentes. Esses aumentos de volatilidade sugerem que o mercado está direcionando maior atenção à ação, possivelmente em resposta a eventos ou expectativas relevantes.

#### 3. Gráfico de *candlestick*
1. Selecione o período dos últimos 60 dias (corridos) para um dos ativos da sua base
2. Monte um gráfico de *candlestick* para esses dados.

````python
df.columns
````

````python
lista_de_indicadores = ['Close', 'High','Low','Open', 'Volatilidade', 'Volume']
ativo = ['IBM'] * len(lista_de_indicadores)
colunas = list(zip(lista_de_indicadores, ativo))
````

````python
df.tail(60)
````

````python
df_ibm = df.loc['2025-01-24':, colunas]
df_ibm.shape
````

````python
df_ibm_viz = df_ibm.stack().reset_index()
````

````python
graph = go.Candlestick(
    x = df_ibm_viz.index,
    open = df_ibm_viz.Open,
    close = df_ibm_viz.Close,
    high = df_ibm_viz.High,
    low = df_ibm_viz.Low,
    name = "IBM",
    showlegend = True,
)
````

````python
layout = Layout(
    title="Gráfico de Candlestick - IBM",
    paper_bgcolor="rgba(0,0,0,0)",
    plot_bgcolor="rgba(0,0,0,0)",
)

grafico = go.Figure(data=[graph], layout=layout)
grafico.update_xaxes(showgrid=True, gridcolor="lightgrey", gridwidth=1)
grafico.update_yaxes(showgrid=True, gridcolor="lightgrey", gridwidth=1)

grafico
````
