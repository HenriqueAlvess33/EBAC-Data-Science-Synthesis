### 1. Carregar as bases

Vamos utilizar nesta tarefa as bases de reconhecimento de atividade humana através do celular. Carregue novamente as bases salvas na tarefa I.

```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.model_selection import GridSearchCV

features_path = './input/features.txt'
subject_train_path = './input/train/subject_train.txt'
x_train_path = './input/train/X_train.txt'
y_train_path = './input/train/y_train.txt'

x_test_path = './input/test/X_test.txt'
y_test_path = './input/test/y_test.txt' 
subject_test_path= './input/test/subject_test.txt'

features = pd.read_csv(features_path, delimiter = '\t', header=None)[0]

subject_train = pd.read_csv(subject_train_path, delimiter = '\t', header=None)[0]
x_train = pd.read_csv(x_test_path, delim_whitespace=True, header=None, names=features)
y_train = pd.read_csv(y_test_path, delim_whitespace=True, header=None)

subject_test = pd.read_csv(subject_test_path, delimiter = '\t', header=None)[0]
x_test = pd.read_csv(x_test_path, delim_whitespace=True, header=None, names=features)
y_test = pd.read_csv(y_test_path, delim_whitespace=True, header=None)


dataframe_treino = (
    pd.DataFrame(x_train)  
    .assign(Indivíduos=subject_train)  
    .assign(Atividades=y_train)  
    .set_index("Indivíduos", append=True)  
)

dataframe_teste = (
    pd.DataFrame(x_test)  
    .assign(Indivíduos=subject_test)  
    .assign(Atividades=y_test)  
    .set_index("Indivíduos", append=True)  
)

pasta = './input/Planilhas/'
os.makedirs(pasta, exist_ok=True)
dataframe_treino.to_csv(pasta + 'Dataframe Treino.csv')
dataframe_teste.to_csv(pasta + 'Dataframe Teste.csv')
```
