### <span style='color: lightblue'>**Fórmula de Bhaskara** </span><br>

A solução para uma função de segundo grau depende das suas raízes (valores de X). Como já vimos, os coeficientes<br> precisam ser números reais e o angular diferente de zero. Sendo assim, temos a seguinte fórmula de Bhaskara:

$$ 
x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a} 
$$

<div align="center"> 

*Fórmula de Bhaskara*

</div>

Para melhorar compreensão o seu cálculo é dividido em duas partes: discriminante da equação e operações para <br> determinar as raízes.

**Discriminante**

A figura dentro da raiz na fórmula de Bhaskara é nomeada de discriminante. Seu símbolo é a letra grega delta e <br>
apresenta a determinada fórmula:

$$ 
\Delta = b^2 - 4ac 
$$

<div align="center">

*Fórmula de discriminante*

</div>

- Se delta for maior que zero, a equação terá dois valores reais e distintos.
- Se o delta for igual a zero, a equação terá somente um valor real e dois resultados iguais.
- Se o delta for menor que zero, a equação na possuirá valores reais
